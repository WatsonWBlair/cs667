## 2.4 EDA Key Findings

*Summarize the most important insights:*

1. **Strongest predictors of mortality**:
    - The strongest prideictor of mortality is a short time period between medical checkups. This is most likely dude to individuals with more health issues seeking more medical intervention. We found that it was prudent to discard this feature as too predictive of the outcome, and would not be applicable to real-world predictive tasks.
    - The next most predictive feature is Serum_Creatinine and Ejection_Fraction. 
        - Serum Creatinine (SCr) levels often rise in response to heart failure (HF), reflecting reduced kidney function due to poor heart output and congestion, a condition known as worsening renal function (WRF) or cardiorenal syndrome; while higher SCr and its increase during hospitalization generally signal worse outcomes, small rises (e.g., <0.3-0.5 mg/dL) can be an expected, acceptable part of effective decongestive therapy with diuretics or neurohormonal blockers, so context matters more than just the number to avoid stopping beneficial treatments
        - Ejection Fraction (EF) is the percentage of blood the heart's left ventricle pumps out with each beat, a key measure of heart function, normally 50-70%; a low EF (under 40%) often signals heart failure or damage from conditions like heart attacks, while a high EF (over 75%) can indicate hypertrophic cardiomyopathy.

2. **Class imbalance**:

# 5. Model Comparison (Task 4)

Comparing predictions from all four models on the same examples.

```python
# Prepare scaled data for LR
X_test_scaled = lr_scaler.transform(X_test)

def get_all_predictions(idx, true_label):
    """Get predictions from all 4 models for a given example."""
    pos = list(X_test.index).index(idx)
    sample = X_test.iloc[pos:pos+1]
    sample_scaled = X_test_scaled[pos:pos+1]
    
    results = {
        'True Label': true_label,
        'LR': lr_model.predict_proba(sample_scaled)[0],
        'DT': dt_model.predict_proba(sample)[0],
        'RF': rf_model.predict_proba(sample)[0],
        'XGB': xgb_model.predict_proba(sample)[0]
    }
    return results
```

## 5.1 Positive Example (True Label: 1 - Death)

```python
# TODO: Fill in POSITIVE_EXAMPLE_IDX
if POSITIVE_EXAMPLE_IDX is not None:
    pos_results = get_all_predictions(POSITIVE_EXAMPLE_IDX, 1)
    
    print(f"True Label: {pos_results['True Label']}")
    print("-" * 40)
    print(f"* LR:  [P(0)={pos_results['LR'][0]:.4f}  P(1)={pos_results['LR'][1]:.4f}]")
    print(f"* DT:  [P(0)={pos_results['DT'][0]:.4f}  P(1)={pos_results['DT'][1]:.4f}]")
    print(f"* RF:  [P(0)={pos_results['RF'][0]:.4f}  P(1)={pos_results['RF'][1]:.4f}]")
    print(f"* XGB: [P(0)={pos_results['XGB'][0]:.4f}  P(1)={pos_results['XGB'][1]:.4f}]")
```

## 5.2 Negative Example (True Label: 0 - Survived)

```python
# TODO: Fill in NEGATIVE_EXAMPLE_IDX
if NEGATIVE_EXAMPLE_IDX is not None:
    neg_results = get_all_predictions(NEGATIVE_EXAMPLE_IDX, 0)
    
    print(f"True Label: {neg_results['True Label']}")
    print("-" * 40)
    print(f"* LR:  [P(0)={neg_results['LR'][0]:.4f}  P(1)={neg_results['LR'][1]:.4f}]")
    print(f"* DT:  [P(0)={neg_results['DT'][0]:.4f}  P(1)={neg_results['DT'][1]:.4f}]")
    print(f"* RF:  [P(0)={neg_results['RF'][0]:.4f}  P(1)={neg_results['RF'][1]:.4f}]")
    print(f"* XGB: [P(0)={neg_results['XGB'][0]:.4f}  P(1)={neg_results['XGB'][1]:.4f}]")
```
