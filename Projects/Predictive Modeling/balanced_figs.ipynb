```python
import pandas as pd

import matplotlib.pyplot as plt

# Load engineered data for modeling
modelingData = pd.read_csv('./data/eng_data.csv')
```

```python
print(modelingData.head().to_markdown())
```

```python
fig, axs = plt.subplots(2, 4, figsize=(10, 5)) 
axs = axs.flatten()
for i, cat in enumerate(['store_region', 'category','payment_method','age_groups','year','month','day','discount']):
    grouped = modelingData.groupby(cat)['total_amount'].sum()
    grouped.plot(kind='pie', ax=axs[i])
    axs[i].set_ylabel('')
    axs[i].set_title(cat.title())

fig.suptitle('Share of total_amount by Grouping', fontsize=16)
plt.show()
```

```python
fig, axs = plt.subplots(2, 4, figsize=(10, 5)) 
axs = axs.flatten()
for i, cat in enumerate(['store_region', 'category','payment_method','age_groups','year','month','day','discount']):
    grouped = modelingData.groupby(cat)['age'].count()
    grouped.plot(kind='pie', ax=axs[i])
    axs[i].set_ylabel('')
    axs[i].set_title(cat.title())

fig.suptitle('Population Size by Grouping', fontsize=16)
plt.show()
```

```python
byRegion = modelingData.groupby(['store_region','payment_method','category','online_or_offline'])['total_amount'].sum()
byRegion.plot.pie(autopct='%.1f%%', figsize=(20, 15), subplots=True)
# 'store_region','payment_method','category','online_or_offline'
plt.title('Target Value Sum when split by Region, Payment Method, Category, and Online_or_Offline')
plt.show()
```
