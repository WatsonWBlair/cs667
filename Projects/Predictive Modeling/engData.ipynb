```python
import pandas as pd
```

```python
cleanData = pd.read_csv('./data/clean_data.csv')
```

```python
print(cleanData.head().to_markdown())
```

```python
print(cleanData['date'].min())
for cat in cleanData['category'].unique():
    # print(cat,':',len(cleanData[cleanData['category'] == cat]))
    print(cat,':',cleanData[cleanData['category'] == cat]['total_amount'].sum())
    print(cat,'quantity:',cleanData[cleanData['category'] == cat]['quantity'].sum())

print('\n')

for reg in cleanData['store_region'].unique():
    # print(cat,':',len(cleanData[cleanData['category'] == cat]))
    print(reg,':',cleanData[cleanData['store_region'] == reg]['total_amount'].sum())

print('\n')
for met in cleanData['payment_method'].unique():
    # print(cat,':',len(cleanData[cleanData['category'] == cat]))
    print(met,':',cleanData[cleanData['payment_method'] == met]['total_amount'].sum())


print('\n')
for ol in cleanData['online_or_offline'].unique():
    # print(cat,':',len(cleanData[cleanData['category'] == cat]))
    print(ol,'sum:',cleanData[cleanData['online_or_offline'] == ol]['total_amount'].sum())
    print(ol,'median:',cleanData[cleanData['online_or_offline'] == ol]['total_amount'].median())
```

```python
engData = cleanData.copy()
```

```python
engData['year'] = pd.to_datetime(engData['date']).dt.year
engData['month'] = pd.to_datetime(engData['date']).dt.month
engData['day'] = pd.to_datetime(engData['date']).dt.day
# engData['day'] = pd.to_datetime(engData['date']).dt.day_of_week
engData = engData.drop(columns=['date'])
```

```python

print('\n')
for yea in engData['year'].unique():
    # print(cat,':',len(cleanData[cleanData['category'] == cat]))
    print(yea,':',engData[engData['year'] == yea]['total_amount'].mean())
    print(yea,':',engData[engData['year'] == yea]['total_amount'].median())
    print(yea,':',engData[engData['year'] == yea]['total_amount'].sum())


print('\n')
for mo in engData['month'].unique():
    # print(cat,':',len(cleanData[cleanData['category'] == cat]))
    print(mo,':',engData[engData['month'] == mo]['total_amount'].mean())
    print(mo,':',engData[engData['month'] == mo]['total_amount'].median())
    print(mo,':',engData[engData['month'] == mo]['total_amount'].sum())

print('\n')
for day in engData['day'].unique():
    # print(cat,':',len(cleanData[cleanData['category'] == cat]))
    print(day,':',engData[engData['day'] == day]['total_amount'].mean())
    print(day,':',engData[engData['day'] == day]['total_amount'].median())
    print(day,':',engData[engData['day'] == day]['total_amount'].sum())
```
