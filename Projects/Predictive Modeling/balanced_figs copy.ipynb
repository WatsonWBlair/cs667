```python
import pandas as pd

import matplotlib.pyplot as plt

# Load engineered data for modeling
modelingData = pd.read_csv('./data/eng_data.csv')
```

```python
print(modelingData.head().to_markdown())
```

```python
byRegion = modelingData.groupby(['store_region','payment_method','category'])
print(byRegion.count())
```

```python
byRegion.plot(kind='bar', figsize=(8, 5))

plt.title('Population Counts when split by Region, Payment Method,  of Value by Category')
plt.xlabel('Category')
plt.ylabel('Sum of Value')
plt.grid(axis='y', linestyle='--')
plt.show()
```
