```python
# configObj = {
#     'OneHot': ['col', 'to', 'onehot'],
#     'target': [('target_col',['col', 'to', 'encode'])],
#     'toOrdinal': [('col', fnToApply)]
# }

defaultConfig = {
    'OneHot': [],
    'target': [],
    'toOrdinal': []
}

def encoder(data: pd.DataFrame, configObj = defaultConfig):
    result = data.copy()

    toOneHot = configObj['OneHot']
    toOrdinal = configObj['ordinal']
    toTarget = configObj['target']
    
    result = pd.get_dummies(result, columns=toOneHot, drop_first=True)

    for [col, func] in toOrdinal:
        result[col] = result.apply(func, axis=1)

    for [target, toEncode] in toTarget:
        for encodeCol in toEncode:
            result[encodeCol] = result.groupby(encodeCol)[target].mean()

```
