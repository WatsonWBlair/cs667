```python
import pandas as pd
```

```python
cleanData = pd.read_csv('./clean_data.csv')
print(len(cleanData))
```

```python
engData = cleanData.copy()

print(engData.head().to_markdown())
```

```python
# Add Time Destruction
engData['Timestamp'] = pd.to_datetime(engData['Timestamp'])

engData['Year'] = engData['Timestamp'].dt.year
engData['Quarter'] = engData['Timestamp'].dt.quarter
engData['Month'] = engData['Timestamp'].dt.month

engData['Day_Of_Week'] = engData['Timestamp'].dt.day_name()
engData['Day_Of_Month'] = engData['Timestamp'].dt.day
engData['Day_Of_Year'] = engData['Timestamp'].dt.day_of_year

engData['Hour'] = engData['Timestamp'].dt.hour
engData['Minute'] = engData['Timestamp'].dt.minute

print(engData.head().to_markdown())
```

```python
engData.to_csv('./eng_data.csv', index=False)
```
