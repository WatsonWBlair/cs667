```python
import pandas as pd
```

```python
rawData = pd.read_excel('./data/raw_data.xlsx')

print(rawData.head().to_markdown())
```

```python
cleanData = rawData.copy()
```

```python
# Remove prefix strings from Account, Transaction and Merchant IDs
cleanData['AccountID'] = cleanData['AccountID'].str.removeprefix('ACC')
cleanData['TransactionID'] = cleanData['TransactionID'].str.removeprefix('TXN')
cleanData['Merchant'] = cleanData['Merchant'].str.removeprefix('Merchant')
print(cleanData.head().to_markdown())
```

```python
cleanData.describe()
```

```python
cleanData.info()
```

```python
# List Unique Categorical Values
print('Unique Values:')
for col in ['AccountID', 'Merchant', 'TransactionType', 'Location']:
    print(col,':',len(cleanData[col].unique()),'->',cleanData[col].unique())
```

```python
# Show range of Timestamp field
cleanData['Timestamp'] = pd.to_datetime(cleanData['Timestamp'])
print(cleanData['Timestamp'].min())
print(cleanData['Timestamp'].max())
```

```python
cleanData.to_csv('./data/clean_data.csv', index=False)
```
